### Testing with simple spring

```python
import sympy as sp 
from sympy import *
```

```python
t, M, l, k = symbols("t M l k",positive=True)
x = Function("x")(t)
x_dot = sp.Derivative(x,t)
```

```python
## Kinetic Energy
T = (1/2)*M*(x_dot**2)

## Potential Energy
k_wall = 1e7
V = Piecewise(
    ((1/2)*k*((l-x)**2) + (1/2)*k_wall*(x**2),x<0),
    ((1/2)*k*((l-x)**2),True)
)

# THE LAGRANGIAN
Lg = T - V
```

```python
x_eqn = Eq(Derivative(Lg,x) - diff((Derivative(Lg,x_dot)),t),0)

x_values = (x_eqn.subs(M,100).subs(k,5).subs(l,10)).doit()

dsolve(solve((x_values.doit()),x))
```

### Coupled Springs

```python
k, l, M, t = sp.symbols("k l M t")
x1 = sp.Function("x1")(t)
x2 = sp.Function("x2")(t)
```

```python
x1_dot = diff(x1,t)
x2_dot = diff(x2,t)
```

```python
epsilon = (10**(-3))*k*(l**4)

epsilon_wall = (10**(-4))*k*(l**4)
```

```python
## Kinetic Energy
T = ((1/2)*M*(x1_dot**2)) + ((1/2)*M*(x2_dot**2))

## Potential Energy
V1 = (1/2)*k*((x1-l)**2) + ((epsilon_wall)*(1/(x1**2))) + ((epsilon)*(1/(x1**2))) 
V2 = (1/2)*k*((x2-x1-l)**2) + ((epsilon_wall)*(1/(x2**2))) + ((epsilon)*(1/((x2-x1)**2)))

V = V1 + V2

## THE LAGRANGIAN
Lg = T - V
```

```python
x1_eqn = Eq(((Derivative(Lg,x1) - diff((Derivative(Lg,x1_dot)),t)).doit()).simplify(),0)
x2_eqn = Eq(((Derivative(Lg,x2) - diff((Derivative(Lg,x2_dot)),t)).doit()).simplify(),0)

sol = solve([x1_eqn,x2_eqn],(diff(x1_dot,t),diff(x2_dot,t)))
```

```python
print(latex(Eq(diff(x1_dot,t),(sol[diff(x1_dot,t)]).simplify())))
```

### THREE BODY

```python
t, M1, M2, M3, R12, R23, R13, G, r = sp.symbols("t M1 M2 M3 R12 R23 R13 G r") 
x1 = sp.Function("x1")(t)
x2 = sp.Function("x2")(t)
x3 = sp.Function("x3")(t)
y1 = sp.Function("y1")(t)
y2 = sp.Function("y2")(t)
y3 = sp.Function("y3")(t)
```

```python
R12 = sp.sqrt(((x1-x2)**2)+((y1-y2)**2))
R23 = sp.sqrt(((x3-x2)**2)+((y3-y2)**2))
R13 = sp.sqrt(((x1-x3)**2)+((y1-y3)**2))
```

```python
x1_dot = diff(x1,t)
x2_dot = diff(x2,t)
x3_dot = diff(x3,t)
y1_dot = diff(y1,t)
y2_dot = diff(y2,t)
y3_dot = diff(y3,t)
```

```python
## Kinetic Energy

T1 = (1/2)*M1*(x1_dot**2 + y1_dot**2)
T2 = (1/2)*M2*(x2_dot**2 + y2_dot**2)
T3 = (1/2)*M3*(x3_dot**2 + y3_dot**2)

T = T1+T2+T3
```

```python
def collision_pot(M1,M2,R12):
    global G,r
    return ((((G*M1*M2)/(2*r))*((0.005*2*r)**6))/((R12-(2*r))**6)).simplify()
```

```python
## Potential Energy
 
V12 = -((G*M1*M2)/R12) + collision_pot(M1,M2,R12)
V13 = -((G*M1*M3)/R13) + collision_pot(M1,M3,R13)
V23 = -((G*M3*M2)/R23) + collision_pot(M3,M2,R23)

V = V12 + V13 + V23
```

```python
## THE LAGRANGE

Lg = T - V
```

```python
def euler_lagrange(Lg,q,q_dot):
    return (Derivative(Lg,q)-diff((Derivative(Lg,q_dot)),t)).doit()
```

```python
q_x = [
    x1,x2,x3
]
q_y = [
    y1,y2,y3
]
q_dot_x = [
    x1_dot,x2_dot,x3_dot
]
q_dot_y = [
    y1_dot,y2_dot,y3_dot
]
```

```python
Eqn_x = [euler_lagrange(Lg,q_x[i],q_dot_x[i]) for i in range(3)]
Eqn_y = [euler_lagrange(Lg,q_y[i],q_dot_y[i]) for i in range(3)]
```

```python
q_ddot_x = [
    diff(x1_dot,t),diff(x2_dot,t),diff(x3_dot,t)
]
q_ddot_y = [
    diff(y1_dot,t),diff(y2_dot,t),diff(y3_dot,t)
]
```

```python
sol = solve(Eqn_x + Eqn_y, q_ddot_x + q_ddot_y, simplify=False)
```
