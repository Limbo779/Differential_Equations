```python
import sympy as sp
from sympy import *
```

```python
x1, y1, x2, y2 = sp.symbols("x1 y1 x2 y2")
t = sp.symbols("t")
L1, L2 = sp.symbols("L1 L2")
theta1 = sp.Function("theta1")(t)
theta2 = sp.Function("theta2")(t)

M1, M2 = sp.symbols("M1 M2")
T1, T2 = sp.symbols("T1 T2")
Lg, T, V = sp.symbols("Lg T V")

g = sp.symbols("g", positive=True)
```

```python
y1 = -L1*cos(theta1)
x1 = L1*sin(theta1)
y2 = -L1*cos(theta1) - L2*cos(theta2)
x2 = L1*sin(theta1) + L2*sin(theta2)
```

```python
y1_dot = y1.diff(t)
x1_dot = x1.diff(t)
y2_dot = y2.diff(t)
x2_dot = x2.diff(t)

theta1_dot = theta1.diff(t)
theta2_dot = theta2.diff(t)
```

```python
## Total Kinetic Energy
T = sp.Rational(1, 2)*((M1*((x1_dot**2)+(y1_dot**2)))+(M2*((x2_dot**2)+(y2_dot**2))))
#T = (1/2)*((M1*((x1_dot**2)+(y1_dot**2)))+(M2*((x2_dot**2)+(y2_dot**2))))

## Total Potential Energy
V = (-1*(M1*g*L1*cos(theta1))) + (-1*(M2*g*((L1*cos(theta1))+(L2*cos(theta2)))))

## THE LAGRANGIAN

Lg = T - V

theta1_eqn = Eq(((Derivative(Lg, theta1) - diff((Derivative(Lg, theta1_dot)),t)).doit()).simplify(),0)
theta2_eqn = Eq(((Derivative(Lg, theta2) - diff((Derivative(Lg, theta2_dot)),t)).doit()).simplify(),0)

sol = solve([theta1_eqn, theta2_eqn],(diff(theta1_dot,t),diff(theta2_dot,t)))

#for i in sol:
#    print(latex(i),end="\n\n")
#
#sol[diff(theta2,t,1)]
```
